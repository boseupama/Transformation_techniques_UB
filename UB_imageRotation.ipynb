```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/pandaUB.sample.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimention
h, w = image.shape[:2]    #height, width, channel
```

```python
#Rotation matrix
Center = (w / 2, h / 2)
Angle = 45        # degrees, positive
Scale = 1.0        # to keep original size

M = cv2.getRotationMatrix2D(Center, Angle, Scale)
```

```python
Rotated = cv2.warpAffine(image, M, (w, h))
```

```python
plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(Rotated)
plt.title('Rotated')
plt.axis('off')

plt.show()
```
