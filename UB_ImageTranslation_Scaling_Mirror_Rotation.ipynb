```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/pandaUB.sample.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimention
h, w = image.shape[:2]    #height, width, channel
```

```python
#translation offsets
tx = 50
ty = 30
```

```python
#Translation matrix
M_translate = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])

translated = cv2.warpAffine(image, M_translate, (w, h))
```

```python
#Scaling factors
scale_x = 3
scale_y = 1.5
```

```python
#Scaling
new_w = int(w * scale_x)
new_h = int(h * scale_y)

scaled = cv2.resize(image, (new_w, new_h), interpolation=cv2.INTER_LINEAR)
```

```python
#Applying mirror
# 1 = horizontal flip (mirror left-right)
# 0 = vertical flip (mirror top-bottom)
# -1 = both
mirrored = cv2.flip(image, 1)
```

```python
#Rotation parameters
center = (w / 2, h / 2)
angle = 45        # degrees, positive = counter-clockwise
scale = 1.0
```

```python
M_rotate = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(image, M_rotate, (w, h))
```

```python
plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(translated)
plt.title('Translated')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(scaled)
plt.title('Scaled')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(mirrored)
plt.title('Mirrored')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(rotated)
plt.title('Rotated')
plt.axis('off')

plt.tight_layout()
plt.show()
```
