```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/mango_back.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimention
h, w = image.shape[:2]     #height, width, channel
```

```python
tx = 50
ty = 30
```

```python
def my_function(tx, ty):
    print("tx =", tx)
    print("ty =", ty)

tx = int(input("Enter tx: "))
ty = int(input("Enter ty: "))

my_function(tx, ty)
```

```python
#translation matrix
M= np.float32([
    [1,0,tx],
    [0,1,ty]
])
```

```python
translated = cv2.warpAffine(image,M, (w,h))
```

```python
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')


plt.subplot(1,2,2)
plt.imshow(translated)
plt.title('Translated')
plt.axis('off')

plt.show()
```
