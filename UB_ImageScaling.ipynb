```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/mango_back.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimention
h, w = image.shape[:2]    #height, width, channel
```

```python
#scaling factors
scale_x = 2   # scale width by 1.5x
scale_y = 1.5   # scale height by 1.5x
```

```python
#computing new dimention
new_w = int(w * scale_x)
new_h = int(h * scale_y)
```

```python
scaled = cv2.resize(image, (new_w, new_h), interpolation=cv2.INTER_LINEAR)
```

```python
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(scaled)
plt.title('Scaled')
plt.axis('off')

plt.show()
```
